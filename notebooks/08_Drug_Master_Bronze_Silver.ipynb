```python
df_drugs_bronze = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/workspace/default/raw_data/drug_master_raw.csv")
)

display(df_drugs_bronze)
```

```python
df_drugs_silver = df_drugs_bronze
```

```python
display(df_drugs_silver)
```

```python
from pyspark.sql.functions import trim, col

df_drugs_silver = (
    df_drugs_silver
    .withColumn("drug_name", trim(col("drug_name")))
    .withColumn("generic_name", trim(col("generic_name")))
    .withColumn("brand_name", trim(col("brand_name")))
    .withColumn("drug_category", trim(col("drug_category")))
    .withColumn("dosage_form", trim(col("dosage_form")))
    .withColumn("unit_of_measure", trim(col("unit_of_measure")))
    .withColumn("prescription_required", trim(col("prescription_required")))
    .withColumn("storage_condition", trim(col("storage_condition")))
)
```

```python
from pyspark.sql.functions import current_timestamp

df_drugs_silver = df_drugs_silver.withColumn(
    "ingestion_timestamp",
    current_timestamp()
)
```

```python
df_drugs_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.pharma_silver.silver_drugs")
```
