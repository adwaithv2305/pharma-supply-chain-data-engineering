```python
df_inventory_bronze = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/workspace/default/raw_data/inventory_raw.csv")
)

display(df_inventory_bronze)
```

```python
df_inventory_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.pharma_bronze.bronze_inventory")
```

```python
from pyspark.sql.functions import current_timestamp

df_inventory_silver = (
    df_inventory_bronze
    .withColumn("ingestion_timestamp", current_timestamp())
)

display(df_inventory_silver)
```

```python
df_inventory_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.pharma_silver.silver_inventory")
```
