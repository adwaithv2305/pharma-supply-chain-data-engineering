```python
df_suppliers_bronze = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/workspace/default/raw_data/supplier_master_raw.csv")
)

display(df_suppliers_bronze)
```

```python
df_suppliers_silver = df_suppliers_bronze
```

```python
display(df_suppliers_silver)
```

```python
df_suppliers_silver = df_suppliers_silver.dropna(how="all")
```

```python
display(df_suppliers_silver)
```

```python
from pyspark.sql.functions import trim, col

df_suppliers_silver = (
    df_suppliers_silver
    .withColumn("supplier_name", trim(col("supplier_name")))
    .withColumn("supplier_type", trim(col("supplier_type")))
    .withColumn("country", trim(col("country")))
    .withColumn("city", trim(col("city")))
    .withColumn("contact_email", trim(col("contact_email")))
    .withColumn("payment_terms", trim(col("payment_terms")))
    .withColumn("supplier_status", trim(col("supplier_status")))
)
```

```python
display(df_suppliers_silver)
```

```python
from pyspark.sql.functions import current_timestamp

df_suppliers_silver = df_suppliers_silver.withColumn(
    "ingestion_timestamp",
    current_timestamp()
)
```

```python
df_suppliers_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.pharma_silver.silver_suppliers")
```
