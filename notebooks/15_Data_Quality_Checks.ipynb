```python
from pyspark.sql.functions import col
```

```python
# Load the Silver tables

po = spark.table("workspace.pharma_silver.silver_purchase_orders")
sales = spark.table("workspace.pharma_silver.silver_sales")
shipments = spark.table("workspace.pharma_silver.silver_shipments")
quality = spark.table("workspace.pharma_silver.silver_quality_events")

print("All Silver tables loaded successfully.")
```

```python
# Purchase Order Data Quality Checks

po_null_ids = po.filter(
    col("po_id").isNull()
).count()

po_duplicate_ids = (
    po.groupBy("po_id")
    .count()
    .filter(col("count") > 1)
    .count()
)

po_negative_quantity = po.filter(
    col("quantity_ordered") < 0
).count()

print("PO NULL IDs:", po_null_ids)
print("PO Duplicate IDs:", po_duplicate_ids)
print("PO Negative Quantities:", po_negative_quantity)
```

```python
# Sales Data Quality Checks

sales_null_ids = sales.filter(
    col("sale_id").isNull()
).count()

sales_duplicate_ids = (
    sales.groupBy("sale_id")
    .count()
    .filter(col("count") > 1)
    .count()
)

sales_negative_quantity = sales.filter(
    col("quantity_sold") < 0
).count()

print("Sales NULL IDs:", sales_null_ids)
print("Sales Duplicate IDs:", sales_duplicate_ids)
print("Sales Negative Quantities:", sales_negative_quantity)
```

```python
# Sales Data Quality Checks

sales_null_ids = sales.filter(
    col("sale_id").isNull()
).count()

sales_duplicate_ids = (
    sales.groupBy("sale_id")
    .count()
    .filter(col("count") > 1)
    .count()
)

sales_negative_quantity = sales.filter(
    col("quantity_sold") < 0
).count()

print("Sales NULL IDs:", sales_null_ids)
print("Sales Duplicate IDs:", sales_duplicate_ids)
print("Sales Negative Quantities:", sales_negative_quantity)    
```

```python
# Shipments Data Quality Checks

shipments_null_ids = shipments.filter(
    col("shipment_id").isNull()
).count()

shipments_duplicate_ids = (
    shipments.groupBy("shipment_id")
    .count()
    .filter(col("count") > 1)
    .count()
)

shipments_negative_quantity = shipments.filter(
    col("quantity_shipped") < 0
).count()

print("Shipments NULL IDs:", shipments_null_ids)
print("Shipments Duplicate IDs:", shipments_duplicate_ids)
print("Shipments Negative Quantities:", shipments_negative_quantity)
```

```python
# Quality Events Data Quality Checks

quality_null_ids = quality.filter(
    col("quality_event_id").isNull()
).count()

quality_duplicate_ids = (
    quality.groupBy("quality_event_id")
    .count()
    .filter(col("count") > 1)
    .count()
)

valid_event_statuses = [
    "Resolved",
    "Closed",
    "Open",
    "Investigating"
]

quality_invalid_status = quality.filter(
    ~col("event_status").isin(valid_event_statuses)
).count()

print("Quality Events NULL IDs:", quality_null_ids)
print("Quality Events Duplicate IDs:", quality_duplicate_ids)
print("Quality Events Invalid Statuses:", quality_invalid_status)
```

```python
# Overall Data Quality Result

dq_results = {
    "Purchase Orders": (
        po_null_ids == 0
        and po_duplicate_ids == 0
        and po_negative_quantity == 0
    ),
    "Sales": (
        sales_null_ids == 0
        and sales_duplicate_ids == 0
        and sales_negative_quantity == 0
    ),
    "Shipments": (
        shipments_null_ids == 0
        and shipments_duplicate_ids == 0
        and shipments_negative_quantity == 0
    ),
    "Quality Events": (
        quality_null_ids == 0
        and quality_duplicate_ids == 0
        and quality_invalid_status == 0
    )
}

for table_name, passed in dq_results.items():
    print(f"{table_name}: {'PASS' if passed else 'FAIL'}")

overall_status = all(dq_results.values())

if overall_status:
    print("Overall Data Quality: PASS")
else:
    raise Exception("Data Quality Check Failed")
```
