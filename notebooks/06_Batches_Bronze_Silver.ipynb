```python
df_batches_bronze = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv("/Volumes/workspace/default/raw_data/batches_raw.csv")
)

display(df_batches_bronze)
```

```python
df_batches_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.pharma_bronze.bronze_batches")
```

```python
from pyspark.sql.functions import current_timestamp

df_batches_silver = (
    df_batches_bronze
    .withColumn("ingestion_timestamp", current_timestamp())
)

display(df_batches_silver)
```

```python
df_batches_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.pharma_silver.silver_batches")
```
